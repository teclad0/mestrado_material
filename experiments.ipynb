```python
import pandas as pd
import plotly.express as px
```

# CORA

```python
cora = pd.read_csv('experiment_results/cora_final_results.csv')
cora
```

```python
cora.columns
```

```python
grouped = cora.groupby('num_particles')['step1_f1_score'].agg(['mean', 'std']).reset_index()

fig = px.line(
    grouped,
    x='num_particles',
    y='mean',
    error_y='std',
    markers=True,
    title='Step1 F1 Score by Number of Particles'
)

fig.update_layout(
    xaxis_title='num_particles',
    yaxis_title='Mean step1_f1_score'
)

fig.show()
```

### coverage

```python
cora['coverage'].describe()
```

### avg_threshold_criteria

```python
cora['avg_threshold_criteria'].value_counts()
```

```python
cora[cora['avg_threshold_criteria'] == True]
```

```python
cora[cora['avg_threshold_criteria'] == True]['avg_node_pot_threshold'].unique()
```

```python
cora[(cora['avg_threshold_criteria'] == False) & (cora['status'] == 'success')]
```

```python
cora[(cora['avg_threshold_criteria'] == False) & (cora['status'] == 'success')]['avg_node_pot_threshold'].unique()
```

avg_node_pot_threshold is not being used in the end, the algorithm can only go up as much as 0.3, contrary to 0.9 that the paper requires as the convergence criteria. noneless it gets good results without it being relevant, its not a determinant of sucess of the algorithm in this setting

### positive_cluster_threshold

```python
grouped = cora[(cora['cluster_strategy'] == 'percentage') & (cora['status'] == 'success')].\
    groupby('positive_cluster_threshold')['step1_f1_score'].agg(['mean', 'std']).reset_index()

fig = px.line(
    grouped,
    x='positive_cluster_threshold',
    y='mean',
    error_y='std',
    markers=True,
    title='Step1 F1 Score by cluster threshold'
)

fig.update_layout(
    xaxis_title='positive_cluster_threshold',
    yaxis_title='Mean step1_f1_score'
)

fig.show()
```

```python
grouped = (
    cora[
        (cora['cluster_strategy'] == 'percentage')
        & (cora['status'] == 'success')
    ]
    .groupby(['percent_positive', 'positive_cluster_threshold'])['step1_f1_score']
    .agg(['mean', 'std'])
    .reset_index()
    .sort_values(['percent_positive', 'positive_cluster_threshold'])
)

fig = px.line(
    grouped,
    x='positive_cluster_threshold',
    y='mean',
    color='percent_positive',
    markers=True,
    title='Step1 F1 Score by cluster threshold and percent_positive',
    custom_data=['percent_positive', 'std']
)

fig.update_traces(
    hovertemplate=(
        'percent_positive=%{customdata[0]:.2f}<br>'
        'positive_cluster_threshold=%{x:.2f}<br>'
        'mean step1_f1_score=%{y:.4f}<br>'
        'std=%{customdata[1]:.4f}<extra></extra>'
    )
)

fig.update_layout(
    xaxis_title='positive_cluster_threshold',
    yaxis_title='Mean step1_f1_score',
    legend_title='percent_positive'
)

fig.show()
```

```python
cora[(cora['cluster_strategy'] == 'percentage') & (cora['status'] == 'success')].\
    groupby('cluster_strategy')['step1_f1_score'].agg(['mean', 'std']).reset_index()
```

### cluster_strategy = majority

```python
cora[(cora['cluster_strategy'] == 'majority') & (cora['status'] == 'success')].\
    groupby('cluster_strategy')['step1_f1_score'].agg(['mean', 'std']).reset_index()
```

the algorithm can get better f1 score with smaller values of positive_cluster_threshold when compared with using the majority strategy. this means that with very small amounts of positive nodes inside clusters is enough for good results.

```python
df_majority = cora[(cora['cluster_strategy'] == 'majority') & (cora['status'] == 'success')]
grouped = df_majority.groupby(['percent_positive', 'num_particles'])['step1_f1_score'].mean().reset_index()

fig = px.line(
    grouped,
    x='num_particles',
    y='step1_f1_score',
    color='percent_positive',
    markers=True,
    title='Majority strategy: Step1 F1 score behavior across experiments'
)

fig.update_layout(
    xaxis_title='num_particles',
    yaxis_title='Mean step1_f1_score',
    legend_title='percent_positive'
)

fig.show()
```

# Total

```python
cora['dataset'] = 'Cora'
citeseer['dataset'] = 'Citeseer'
pubmed['dataset'] = 'Pubmed'
mnist['dataset'] = 'MNIST'
twitch['dataset'] = 'Twitch'
total = pd.concat([cora, citeseer, pubmed, mnist, twitch], ignore_index=True)
total
```

### positive_cluster_threshold

```python
grouped = total[total['status'] == 'success'].groupby(['dataset', 'positive_cluster_threshold'])['step1_f1_score'].agg(['mean', 'std']).reset_index()
fig = px.line(grouped, x='positive_cluster_threshold', y='mean', color='dataset', markers=True,)
fig.show()
```

### cluster_strategy = majority

```python
import warnings
warnings.filterwarnings('ignore')

import numpy as np
import pandas as pd
import random
import time
import torch
import os
from dataset_system import DatasetManager
from models import RCSVM, CCRNE, PU_LP, LP_PUL, MCLS
from model import PULearningPC
from models_experiment import (
    evaluate_f1_score, evaluate_precision_score, evaluate_phase2, classify_with_svm
)
from experiment_config import BASELINE_CONFIG, DATASET_GEN_PARAMS

# === Configuration ===
DATASETS = ['cora', 'citeseer', 'mnist', 'twitch', 'pubmed']
PERCENT_POSITIVES = [0.01, 0.05, 0.1, 0.15, 0.2, 0.25]
N_RUNS = 5
NUM_NEG = {'cora': 200, 'citeseer': 200, 'mnist': 300, 'twitch': 200, 'pubmed': 200}

# Baseline model params (from colleague\'s paper, Section 4.2, page 57)
MODEL_PARAMS = {
    'RCSVM': {'alpha': 0.7, 'beta': 0.3},
    'CCRNE': {'ratio': 0.3},
    'PU_LP': {'alpha': 0.1, 'm': 3, 'l': 1},
    'MCLS': {'k': 7, 'ratio': 0.1},
    'LP_PUL': {},
}

PULPC_PARAMS = {
    'num_particles': 100,
    'cluster_strategy': 'percentage',
    'positive_cluster_threshold': 0.1,
    'movement_strategy': 'uniform',
    'initialization_strategy': 'random',
    'avg_node_pot_threshold': 0.8,
}

def run_model_on_graph(model_name, graph, manager, num_neg, params):
    """Run a single model (step1 + step2) on a graph."""
    if model_name == 'CP-APNR':
        pulearning = PULearningPC(
            graph,
            num_neg=num_neg,
            pcm_params={
                'num_particles': PULPC_PARAMS['num_particles'],
                'movement_strategy': PULPC_PARAMS['movement_strategy'],
                'initialization_strategy': PULPC_PARAMS['initialization_strategy'],
                'average_node_potential_threshold': PULPC_PARAMS['avg_node_pot_threshold'],
            },
            rns_params={
                'cluster_strategy': PULPC_PARAMS['cluster_strategy'],
                'positive_cluster_threshold': PULPC_PARAMS['positive_cluster_threshold'],
            }
        )
        pulearning.train()
        rn_eval = pulearning.select_reliable_negatives()
    else:
        if model_name == 'RCSVM':
            model = RCSVM(graph, **params)
            needs_mapping, node_list = False, None
        elif model_name == 'CCRNE':
            model = CCRNE(graph, **params)
            needs_mapping, node_list = False, None
        elif model_name == 'PU_LP':
            model = PU_LP(graph, **params)
            needs_mapping, node_list = False, None
        elif model_name == 'MCLS':
            node_list = sorted(list(graph.nodes()))
            mcls_data = manager.get_data_for_mcls(graph)
            model = MCLS(mcls_data, **params)
            needs_mapping = True
        elif model_name == 'LP_PUL':
            node_list = sorted(list(graph.nodes()))
            lp_pul_data = manager.get_data_for_lp_pul(graph)
            model = LP_PUL(lp_pul_data)
            needs_mapping = True

        model.train()
        rn = model.negative_inference(num_neg)
        rn_list = rn.tolist() if isinstance(rn, torch.Tensor) else list(rn)

        if needs_mapping and node_list is not None:
            rn_eval = [node_list[i] for i in rn_list if i < len(node_list)]
        else:
            rn_eval = rn_list

    f1_s1 = evaluate_f1_score(graph, rn_eval, target_negatives=num_neg)
    prec_s1 = evaluate_precision_score(graph, rn_eval, target_negatives=num_neg)
    predictions = classify_with_svm(graph, rn_eval)
    phase2 = evaluate_phase2(graph, predictions)

    return {
        'step1_f1': f1_s1, 'step1_precision': prec_s1,
        'step2_f1': phase2['f1'], 'step2_precision': phase2['precision'],
        'step2_recall': phase2['recall'], 'step2_accuracy': phase2['accuracy'],
    }

# === Generate datasets and run all models sequentially ===
all_models = {**MODEL_PARAMS, 'CP-APNR': {}}
manager = DatasetManager(random_seed=42)
all_results = []

for dataset in DATASETS:
    num_neg = NUM_NEG[dataset]
    print(f"\n{'='*60}")
    print(f"  {dataset.upper()}")
    print(f"{'='*60}")

    for pct in PERCENT_POSITIVES:
        gen_params = DATASET_GEN_PARAMS.get(dataset, {})
        filepath = manager.generate_and_save_dataset(dataset, gen_params, None, pct)
        graph = manager.load_graph(filepath)

        for model_name, params in all_models.items():
            t0 = time.time()
            for run in range(N_RUNS):
                try:
                    res = run_model_on_graph(model_name, graph, manager, num_neg, params)
                    res.update({'model': model_name, 'dataset': dataset,
                                'percent_positive': pct, 'run': run})
                    all_results.append(res)
                except Exception as e:
                    all_results.append({
                        'model': model_name, 'dataset': dataset,
                        'percent_positive': pct, 'run': run,
                        'step1_f1': 0.0, 'step1_precision': 0.0,
                        'step2_f1': 0.0, 'step2_accuracy': 0.0,
                        'step2_precision': 0.0, 'step2_recall': 0.0,
                        'error': str(e)
                    })
            elapsed = time.time() - t0
            print(f"  {pct:.2f} | {model_name:8s}: {elapsed:.1f}s")

# === Save ==='
results_df = pd.DataFrame(all_results)
baseline_df = results_df[results_df['model'] != 'CP-APNR']
pulpc_df = results_df[results_df['model'] == 'CP-APNR']

baseline_df.to_csv('experiment_results/baselines_5runs_full.csv', index=False)
pulpc_df.to_csv('experiment_results/cpapnr_5runs_full.csv', index=False)
print(f"\nDone! Baselines: {len(baseline_df)} rows | CP-APNR: {len(pulpc_df)} rows")
```

```python
# === Generate Comparison Tables ===

# Combine baseline and PULearningPC results
combined = pd.concat([baseline_df, pulpc_df], ignore_index=True)

# Define model order for display
MODEL_ORDER = ['RCSVM', 'CCRNE', 'PU_LP', 'MCLS', 'LP_PUL', 'CP-APNR']

def make_comparison_table(combined_df, metric, datasets, pcts, model_order):
    """Generate one table per dataset: rows=models, cols=percent_positive, values=mean±std."""
    tables = {}
    for dataset in datasets:
        ds_df = combined_df[combined_df['dataset'] == dataset]
        rows = []
        for model in model_order:
            row = {'Model': model}
            for pct in pcts:
                subset = ds_df[(ds_df['model'] == model) & (ds_df['percent_positive'] == pct)]
                valid = subset[subset[metric] > 0][metric]
                if len(valid) > 0:
                    row[f'{pct}'] = f'{valid.mean():.4f} ± {valid.std():.4f}'
                else:
                    row[f'{pct}'] = '—'
            rows.append(row)
        tables[dataset] = pd.DataFrame(rows)
    return tables

# Step 1: F1 Score tables
print("=" * 80)
print("STEP 1: F1 SCORE (Reliable Negative Extraction)")
print("=" * 80)
step1_tables = make_comparison_table(combined, 'step1_f1', DATASETS, PERCENT_POSITIVES, MODEL_ORDER)
for dataset, table in step1_tables.items():
    print(f"\n--- {dataset.upper()} ---")
    print(table.to_string(index=False))

# Step 2: Accuracy tables
print("\n" + "=" * 80)
print("STEP 2: ACCURACY (Binary Classification)")
print("=" * 80)
step2_tables = make_comparison_table(combined, 'step2_accuracy', DATASETS, PERCENT_POSITIVES, MODEL_ORDER)
for dataset, table in step2_tables.items():
    print(f"\n--- {dataset.upper()} ---")
    print(table.to_string(index=False))

# Step 2: F1 Score tables
print("\n" + "=" * 80)
print("STEP 2: F1 SCORE (Binary Classification)")
print("=" * 80)
step2_f1_tables = make_comparison_table(combined, 'step2_f1', DATASETS, PERCENT_POSITIVES, MODEL_ORDER)
for dataset, table in step2_f1_tables.items():
    print(f"\n--- {dataset.upper()} ---")
    print(table.to_string(index=False))
```
